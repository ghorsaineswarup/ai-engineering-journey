```python
x_tf = tf.Variable(25, dtype=tf.int16)
x_tf
```

```python
x_tf.shape
```

```python
y_tf = tf.Variable(3, dtype=tf.int16)
```

```python
tf_sum.numpy()  # note that NumPy operations automatically convert tensors to NumPy arrays, and vice versa
```
